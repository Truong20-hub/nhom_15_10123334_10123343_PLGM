# 📊 BÀI TẬP LỚN MÔN HỌC MÁY CƠ BẢN
## GIAI ĐOẠN 1: KHÁM PHÁ VÀ PHÂN TÍCH DỮ LIỆU (EDA)
**Đề tài:** Phân loại phân khúc giá điện thoại thông minh (Mobile Price Classification)
**Dữ liệu:** Kaggle Mobile Price Dataset (2000 mẫu, 20 đặc trưng, 4 phân khúc giá)

---

```python
# 1. Khai báo các thư viện cần thiết
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cài đặt đồ họa trực quan
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
```

```python
# 2. Nạp dữ liệu từ train.csv
df = pd.read_csv('../data/train.csv')
print(f'Kích thước tập dữ liệu: {df.shape}')
df.head()
```

```python
# 3. Thống kê mô tả và kiểm tra Missing Values
print('Thông tin các kiểu dữ liệu và giá trị thiếu:')
df.info()
print('\nSố lượng giá trị khuyết thiếu trong từng cột:')
print(df.isnull().sum())
```

### 📌 Hình 1: Phân phối nhãn mục tiêu `price_range`
- **Quan sát:** Cả 4 phân khúc đều có đúng 500 mẫu (25%).
- **Ý nghĩa:** Dữ liệu hoàn toàn cân bằng, không bị Class Imbalance.
- **Quyết định:** Sử dụng Accuracy và Macro-F1 làm thước đo chính.

```python
colors = ['#2ec4b6', '#3a86ff', '#8338ec', '#ff006e']
label_names = ['0: Low', '1: Medium', '2: High', '3: Very High']

plt.figure(figsize=(7, 4))
counts = df['price_range'].value_counts().sort_index()
bars = plt.bar(label_names, counts.values, color=colors, width=0.55, edgecolor='black')
plt.title('Phân phối biến mục tiêu (price_range)')
plt.xlabel('Tầm giá')
plt.ylabel('Số lượng mẫu')
plt.show()
```

### 📌 Hình 2 & 3: Tác động của RAM đến phân khúc giá
- **Quan sát:** RAM có tương quan mạnh r=0.917 với giá bán. Phân bố RAM tăng dần theo bậc thang rõ rệt.
- **Ý nghĩa:** RAM là yếu tố quyết định hàng đầu trong giá thành phần cứng smartphone.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(x='price_range', y='ram', data=df, palette=colors, ax=ax1)
ax1.set_title('Biểu đồ Boxplot RAM theo phân khúc')

sns.violinplot(x='price_range', y='ram', data=df, palette=colors, ax=ax2)
ax2.set_title('Biểu đồ Violin Plot RAM theo phân khúc')
plt.tight_layout()
plt.show()
```

### 📌 Hình 4: Ma trận tương quan đặc trưng (Heatmap)

```python
plt.figure(figsize=(14, 10))
corr = df.corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='Blues', linewidths=0.5)
plt.title('Ma trận hệ số tương quan Pearson giữa 21 biến')
plt.show()
```
